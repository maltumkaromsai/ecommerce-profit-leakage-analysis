#Which category is responsible for most of these discounts?

#If Electronics had followed the same discount rate as Beauty (33.6%), how much additional revenue could the company have retained?

```python
electronics_revenue=category_discounts.loc['Electronics','revenue']

electronics_discount=category_discounts.loc['Electronics','discount%']

beauty_discount=category_discounts.loc['Beauty','discount%']

discount_diff=electronics_discount-beauty_discount

savings=electronics_revenue*(discount_diff/100)

print(f'electronics revenue:{electronics_revenue:.2f}')
print(f'electronics discount:{electronics_discount:.2f}')
print(f'beauty discount:{beauty_discount:.2f}')
print(f'discount difference:{discount_diff:.2f}')
print(f'potential retain:{savings:.2f}')
```

# MODULE 9 - PROFIT LEAKAGE FROM OUTSIDE PROMOTION ORDERS

#How much profit did we make/lose from these suspicious outside-promotion orders?

```python
leakage_orders=[]

for _, product in premium_discount.iterrows():
  product_id=product['ProductID']
  product_name=product['ProductName']
  high_discount=product['DiscountApplied']

  orders=merged_df[
      (merged_df['ProductID']==product_id) &
      (merged_df['DiscountApplied']==high_discount)
  ].copy()

  promo=promotions[
      (promotions['ProductID']==product_id) &
      (promotions['DiscountPercent']==high_discount)
  ]

  outside_orders=[]

  for _, order in orders.iterrows():
    order_date=order['OrderDate']

    inside_promotion=any(
        (order_date>=start) and (order_date<=end)
        for start,end in zip(promo['StartDate'],promo['EndDate'])
    )

    if not inside_promotion:
      outside_orders.append(order)

  if outside_orders:
    outside_orders=pd.DataFrame(outside_orders)

    leakage_orders.append({
        'ProductName':product_name,
        'HighDiscount':high_discount,
        'OutsideOrders':len(outside_orders),
        'Quantity':outside_orders['Quantity'].sum(),
        'Revenue':outside_orders['Revenue'].sum(),
        'Profit':outside_orders['Profit'].sum()
    })

leakage_summary=pd.DataFrame(leakage_orders)

leakage_summary=leakage_summary.sort_values(
    'Profit',ascending=True
)

leakage_summary
```
